```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/mnt/Workplace/My Codes/Python/Feature Engineering/Student_Depression_Dataset.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.dropna(inplace=True)
df.drop_duplicates(inplace=True)
```

```python
df.columns
```

```python
columns_to_drop=['Gender', 'Family_background', 'Relationship_status',
       'Marital_status', 'Living_style', 'CGPA',
       'friend_circle', 'Health_condition',
       'Home_division', 'Depression']
```

```python
df.drop(columns=columns_to_drop)
```

```python
X = pd.DataFrame(df, columns=['Age','Cost_without_study'])
Y = pd.DataFrame(df, columns=['Current_semester'])
```

```python
#Train Test Split

from sklearn.model_selection import train_test_split
X_train, X_test, Y_train, Y_test  = train_test_split(X, Y, random_state=0, test_size=0.2)

X_train.shape, X_test.shape
```

```python
#StandardScaler

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
scaler.fit(X_train)

X_train_scaled = scaler.transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
scaler.mean_
```

```python
X_train_scaled = pd.DataFrame(X_train_scaled, columns=X_train.columns)
```

```python
X_train_scaled
```

```python
fig, (ax1, ax2) = plt.subplots(ncols=2, figsize=(12,5))

ax1.scatter(X_train['Age'], X_train['Cost_without_study'])
ax1.set_title("Before Scaling")

ax2.scatter(X_train_scaled['Age'], X_train_scaled['Cost_without_study'])
ax2.set_title("After Scaling")
```

```python
from sklearn.linear_model import LogisticRegression

lr = LogisticRegression()
lr_scaled = LogisticRegression()
```

```python
lr.fit(X_train,Y_train)
lr_scaled.fit(X_train_scaled,Y_train)
```

```python
y_pred = lr.predict(X_test)
y_pred_scaled = lr_scaled.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score

print("Actual: ", accuracy_score(Y_test, y_pred)*100)
print("Scaled: ", accuracy_score(Y_test, y_pred_scaled)*100)
```
